```python
import polars as pl

df = pl.read_parquet("./rbpf/data/results.parquet")
df
```

```python
df.filter(
    pl.col("tournament") == "FIFA World Cup",
    pl.col("date").str.to_date().dt.year() == 2026,
).unpivot(
    on=["home_team", "away_team"],
    variable_name="side",
    value_name="team",
).select(
    pl.col("team").unique().sort()
).to_series().to_list()
```
